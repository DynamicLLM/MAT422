```python
import numpy as np
import matplotlib.pyplot as plt
from collections import deque

np.set_printoptions(precision=3, suppress=True)
```

## Graph basics and adjacency representations

This section gives one compact Python demonstration of **Graph basics and adjacency representations**. After running the code, add your own interpretation and change at least one input, parameter, or example so the notebook reflects your own work.


```python
nodes = ["A", "B", "C", "D", "E"]
edges = [("A","B"), ("A","C"), ("B","D"), ("C","D"), ("D","E")]
index = {node: i for i, node in enumerate(nodes)}
adj = np.zeros((len(nodes), len(nodes)), dtype=int)
for u, v in edges:
    adj[index[u], index[v]] = 1
    adj[index[v], index[u]] = 1

print("Adjacency matrix:")
print(adj)
print("Degrees:", dict(zip(nodes, adj.sum(axis=1))))
```

## Degree and neighbors

This section gives one compact Python demonstration of **Degree and neighbors**. After running the code, add your own interpretation and change at least one input, parameter, or example so the notebook reflects your own work.


```python
adj_list = {node: [] for node in nodes}
for u, v in edges:
    adj_list[u].append(v)
    adj_list[v].append(u)

def bfs(start):
    seen = {start}
    order = []
    queue = deque([start])
    while queue:
        node = queue.popleft()
        order.append(node)
        for neighbor in adj_list[node]:
            if neighbor not in seen:
                seen.add(neighbor)
                queue.append(neighbor)
    return order

print("Adjacency list:", adj_list)
print("BFS order from A:", bfs("A"))
```

## Graph traversal

This section gives one compact Python demonstration of **Graph traversal**. After running the code, add your own interpretation and change at least one input, parameter, or example so the notebook reflects your own work.


```python
positions = {
    "A": (0, 1),
    "B": (1, 2),
    "C": (1, 0),
    "D": (2, 1),
    "E": (3, 1),
}

plt.figure(figsize=(6, 3))
for u, v in edges:
    x = [positions[u][0], positions[v][0]]
    y = [positions[u][1], positions[v][1]]
    plt.plot(x, y, color="gray", linewidth=2)
for node, (x, y) in positions.items():
    plt.scatter(x, y, s=600, color="tab:blue")
    plt.text(x, y, node, color="white", ha="center", va="center", fontsize=14)
plt.axis("off")
plt.title("A graph model of connections")
plt.show()
```

## Graph and graph modeling

This section gives one compact Python demonstration of **Graph and graph modeling**. After running the code, add your own interpretation and change at least one input, parameter, or example so the notebook reflects your own work.


```python
# Example interpretation: a simple contact network.
infection_state = {"A": "infected", "B": "susceptible", "C": "susceptible", "D": "susceptible", "E": "susceptible"}
risk_scores = {}
for node in nodes:
    infected_neighbors = sum(infection_state[n] == "infected" for n in adj_list[node])
    risk_scores[node] = infected_neighbors / max(1, len(adj_list[node]))

print("Risk score from infected neighbors:")
print(risk_scores)
```
